```python
import pandas as pd
import numpy as np
import matplotlib . pyplot as plt
# Create dummy dataset
data = {
    'Title': ['Movie A', 'Movie B', 'Movie C', 'Movie D', 'Movie E', 'Movie F', 'Movie G', 'Movie H', 'Movie I', 'Movie J'],
    'Rating': [8.5, 7.8, 9.1, 6.5, 8.2, 7.5, 8.9, 6.8, 7.9, 9.0],
    'Genre': ['Action, Thriller', 'Comedy, Drama', 'Drama, Romance', 'Action, Sci-Fi', 'Comedy, Romance', 'Drama, Thriller', 'Action, Adventure', 'Comedy, Drama', 'Thriller, Crime', 'Drama, Biography'],
    'Year': [2020, 2019, 2021, 2018, 2020, 2019, 2021, 2020, 2019, 2021],
    'Votes': [125000, 89000, 234000, 45000, 98000, 67000, 189000, 56000, 78000, 210000],
    'Gross': ['$50M', '$35M', '$120M', '$25M', '$45M', '$30M', '$95M', '$28M', '$42M', '$110M'],
    'Runtime': ['148 min', '112 min', '165 min', '98 min', '108 min', '125 min', '152 min', '104 min', '118 min', '158 min'],
    'Director': ['Director1', 'Director2', 'Director3', 'Director2', 'Director1', 'Director3', 'Director1', 'Director2', 'Director3', 'Director1'],
    'Actor': ['Actor A', 'Actor B', 'Actor C', 'Actor A', 'Actor B', 'Actor C', 'Actor A', 'Actor B', 'Actor C', 'Actor A']
}
df = pd . DataFrame ( data )
print (" Dataset created successfully ! ")
print ( f" Shape : { df . shape } ")
```

```python
# Display first 3 rows
print (" First 3 rows : ")
print ( df . head (3) )

# Display last 2 rows
print ("\nLast 2 rows : ")
print ( df . tail (2) )
```

```python
# Get DataFrame information
print (" DataFrame Info : ")
print ( df . info () )
```

```python
# Statistics for numeric columns
print (" Statistical Summary : ")
print ( df . describe () )

# Include all columns including object type
print ("\nAll columns statistics : ")
print ( df . describe ( include ='all' ) )
```

```python
# shape attribute - returns ( rows , columns )
print ( f" Number of rows : { df . shape [0]} ")
print ( f" Number of columns : { df . shape [1]} ")

# columns attribute - returns column names
print ( f" Column names : { df . columns . tolist () } ")

# index attribute - returns row indices
print ( f" Index : { df . index } ")

# dtypes attribute - returns data types
print ( f" Data types :\n { df . dtypes } ")

# values attribute - returns numpy array representation
print ( f" Values as array :\n { df . values [:2]} ")
```

```python
# Remove $ and M from Gross column
df['Gross_Clean'] = df['Gross'].str.replace('$', '', regex=False).str.replace('M', '', regex=False)

print("Cleaned Gross values : ")
print(df[['Gross', 'Gross_Clean']].head())
```

```python
df['Genre_List'] = df['Genre'].str.split(', ')
print("Split genres : ")
print(df[['Genre', 'Genre_List']].head())
```

```python
# Create genre frequency by stacking
all_genres = df['Genre_List'].explode() # Alternative to stack
genre_counts = all_genres.value_counts()
print("Genre frequencies : ")
print(genre_counts)
```

```python
# Convert Gross_Clean to float
df['Gross_Millions'] = df['Gross_Clean'].astype(float)
print(f"Data type after conversion : {df['Gross_Millions'].dtypes}")

# Convert Runtime to integer
df['Runtime_Minutes'] = df['Runtime'].str.replace(' min', '').astype(int)

print(df[['Runtime', 'Runtime_Minutes']].head())
```

```python
# Check for missing values
print("Missing values per column : ")
print(df.isnull().sum())

# Fill missing values
df_filled = df.fillna({'Rating': df['Rating'].mean(), 'Votes': 0})

# Drop rows with any missing values
df_dropped = df.dropna()

# Drop columns with missing values
df_no_na_cols = df.dropna(axis=1)
```

```python
# Sort by Rating descending
df_sorted = df.sort_values('Rating', ascending=False)
print("Top 5 highest rated movies : ")
print(df_sorted[['Title', 'Rating']].head())

# Sort by multiple columns
df_multi_sort = df.sort_values(['Year', 'Rating'], ascending=[False, False])
print("\nSorted by Year (desc) then Rating (desc) : ")
print(df_multi_sort[['Title', 'Year', 'Rating']])
```

```python
 # Reset index and sort by index
df_reset = df . reset_index ()
print (" Sorted by index : ")
print ( df_reset . sort_index ( ascending = False ) . head () )
```

```python
 # Top 3 highest rated movies
top_3 = df . nlargest (3 , 'Rating')
print (" Top 3 movies by rating : ")
print ( top_3 [[ 'Title', 'Rating']])

# Bottom 3 by votes
bottom_3 = df . nsmallest (3 , 'Votes')
print (" \nBottom 3 movies by votes : ")
print ( bottom_3 [[ 'Title', 'Votes']])
```

```python
 # Count movies per director
director_counts = df ['Director'].value_counts()
print (" Movies per director : ")
print ( director_counts )

# Get percentages
director_pct = df ['Director'].value_counts(normalize = True) * 100
print (" \nDirector distribution (%) : ")
print ( director_pct )

# Top 3 most frequent actors
top_actors = df ['Actor'].value_counts() . head (3)
print (" \nTop 3 actors : ")
print ( top_actors )
```

```python
 # Group by single column
grouped_by_director = df . groupby ('Director')
print (" Groups created by director : ")
print ( grouped_by_director . size () ) # Size of each group

# Calculate mean rating per director
mean_rating = df . groupby ('Director') ['Rating']. mean ()
print (" \nAverage rating by director : ")
print ( mean_rating )

# Multiple aggregations
agg_results = df . groupby ('Director') . agg ({
'Rating': ['mean', 'min', 'max', 'count'] ,
'Gross_Millions': 'sum',
'Votes': 'median'
})
print (" \nMultiple aggregations : ")
print ( agg_results )
```

```python
# Apply multiple aggregation functions simultaneously.
 # Using agg with custom functions
def rating_range ( x ) :
 return x . max () - x . min ()

stats = df . groupby ('Year') ['Rating']. agg ([
 ('Average', 'mean') ,
 ('Maximum', 'max') ,
 ('Minimum', 'min') ,
 ('Range', rating_range ) ,
 ('Std_Dev', 'std')
 ])
print (" Yearly rating statistics : ")
print ( stats )

# Different aggregations for different columns
summary = df . groupby ('Year') . agg ({
 'Rating': ['mean', 'std'] ,
 'Gross_Millions': 'sum',
 'Votes': 'count'
 })
print (" \nCombined aggregations : ")
print ( summary )
```

```python
# Rename single column
df_renamed = df . rename ( columns ={ 'Rating' : 'IMDB_Rating' })
print (" Column renamed : ")
print ( df_renamed . columns . tolist () )

# Rename multiple columns
df_multiple = df . rename ( columns ={
'Title' : 'Movie_Title' ,
'Votes' : 'Number_of_Votes' ,
'Gross' : 'Box_Office'
})
print (" \nMultiple columns renamed : ")
print ( df_multiple . columns . tolist () )

# Rename index
df_index_renamed = df . rename ( index ={0: 'First' , 1: 'Second' })
print (" \nIndex renamed : ")
print ( df_index_renamed . head (2) )
```

```python
# Create two subsets
df_2020 = df [ df ['Year'] == 2020]
df_2021 = df [ df ['Year'] == 2021]

# Concatenate vertically ( rows )
df_combined = pd . concat ([ df_2020 , df_2021 ] , ignore_index = True )
print ( f" Combined shape : { df_combined . shape } ")
print (" Combined years : ")
print ( df_combined ['Year']. value_counts () . sort_index () )

# Concatenate horizontally ( columns )
df_extra = pd . DataFrame ({
'Budget_Millions' : [100 , 85 , 120 , 95 , 110 , 88 , 115 , 92 , 105 , 130] ,
'Awards_Won' : [2 , 1 , 5 , 0 , 3 , 1 , 4 , 0 , 2 , 6]
})
df_with_budget = pd . concat ([ df , df_extra ] , axis =1)
print (" \nWith budget column : ")
print ( df_with_budget . columns . tolist () )
```

```python
# Create additional dataframe
director_info = pd . DataFrame ({
'Director' : ['Director1' , 'Director2' , 'Director3' , 'Director4'] ,
'Country' : ['USA' , 'UK' , 'Canada' , 'France'] ,
'Years_Active' : [15 , 10 , 20 , 5]
})

# Inner join ( default )
df_merged = pd . merge ( df , director_info , on ='Director' , how ='inner' )
print (" Inner join result : ")
print ( df_merged [[ 'Title' , 'Director' , 'Country' ]]. head () )

# Left join
df_left = pd . merge ( df , director_info , on ='Director' , how ='left' )
print ( f" \nLeft join shape : { df_left . shape } " )

# Right join
df_right = pd . merge ( df , director_info , on ='Director' , how ='right' )
print ( f" Right join shape : { df_right . shape } " )

# Outer join
df_outer = pd . merge ( df , director_info , on ='Director' , how ='outer' )
print ( f" Outer join shape : { df_outer . shape } " )
```

```python
# Create rating categories
rating_bins = [0 , 7 , 8 , 9 , 10]
rating_labels = ['Poor' , 'Good' , 'Very Good' , 'Excellent']
df ['Rating_Category'] = pd . cut ( df ['Rating'] , bins = rating_bins , labels = rating_labels )

print (" Rating categories : ")
print ( df [[ 'Title' , 'Rating' , 'Rating_Category']])

# Equal - width bins
df ['Rating_Bin'] = pd . cut ( df ['Rating'] , bins =4)
print (" \nEqual width bins : ")
print ( df ['Rating_Bin']. value_counts () )

# Get bin edges
bins = pd . cut ( df ['Rating'] , bins =3 , retbins = True )
print ( f" \nBin edges : { bins [1]} ")
```

```python
import matplotlib.pyplot as plt

# Basic plot configuration
plt.style.use("seaborn-v0_8-darkgrid")
plt.figure(figsize=(12, 8))

# Common plot parameters
plt.xlabel("X-axis Label")
plt.ylabel("Y-axis Label")
plt.title("Plot Title")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()

plt.show()
```

```python
# Simple histogram
plt.figure(figsize=(10, 6))

plt.hist(
    df["Rating"],
    bins=10,
    edgecolor="black",
    alpha=0.7,
    color="skyblue"
)

plt.xlabel("IMDB Rating")
plt.ylabel("Frequency")
plt.title("Distribution of Movie Ratings")
plt.grid(True, alpha=0.3)
plt.show()


# Multiple histograms
plt.figure(figsize=(10, 6))

plt.hist(
    df[df["Year"] == 2020]["Rating"],
    bins=10,
    alpha=0.5,
    label="2020"
)

plt.hist(
    df[df["Year"] == 2021]["Rating"],
    bins=10,
    alpha=0.5,
    label="2021"
)

plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.title("Rating Distribution by Year")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Basic scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(
    df["Votes"],
    df["Rating"],
    alpha=0.6,
    s=100,
    color="green"
)

plt.xlabel("Number of Votes")
plt.ylabel("Rating")
plt.title("Rating vs Number of Votes")
plt.grid(True, alpha=0.3)
plt.show()


# Scatter plot colored by year
plt.figure(figsize=(10, 6))

colors = {
    2020: "red",
    2019: "blue",
    2018: "green",
    2021: "purple"
}

for year in df["Year"].unique():
    year_data = df[df["Year"] == year]

    plt.scatter(
        year_data["Votes"],
        year_data["Rating"],
        label=year,
        alpha=0.6,
        s=100,
        color=colors[year]
    )

plt.xlabel("Votes")
plt.ylabel("Rating")
plt.title("Rating vs Votes by Year")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Vertical bar chart: Average Rating by Director
avg_rating_by_director = df.groupby("Director")["Rating"].mean()

plt.figure(figsize=(10, 6))
plt.bar(
    avg_rating_by_director.index,
    avg_rating_by_director.values,
    color="coral",
    edgecolor="black"
)

plt.xlabel("Director")
plt.ylabel("Average Rating")
plt.title("Average Rating by Director")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3, axis="y")
plt.tight_layout()
plt.show()


# Horizontal bar chart: Total Gross by Director
gross_by_director = df.groupby("Director")["Gross_Millions"].sum()

plt.figure(figsize=(10, 6))
plt.barh(
    gross_by_director.index,
    gross_by_director.values,
    color="lightgreen",
    edgecolor="black"
)

plt.xlabel("Total Gross (Millions USD)")
plt.ylabel("Director")
plt.title("Total Box Office Collection by Director")
plt.grid(True, alpha=0.3, axis="x")
plt.tight_layout()
plt.show()
```

```python
# Rating trend over years
yearly_avg = df.groupby("Year")["Rating"].mean()

plt.figure(figsize=(10, 6))
plt.plot(
    yearly_avg.index,
    yearly_avg.values,
    marker="o",
    linewidth=2,
    markersize=10,
    color="blue"
)

plt.xlabel("Year")
plt.ylabel("Average Rating")
plt.title("Movie Rating Trend Over Years")
plt.xticks(yearly_avg.index)
plt.grid(True, alpha=0.3)
plt.ylim(7, 10)
plt.tight_layout()
plt.show()


# Multiple lines: Rating statistics by year
yearly_stats = df.groupby("Year").agg({
    "Rating": ["mean", "min", "max"]
}).round(2)

plt.figure(figsize=(10, 6))

plt.plot(
    yearly_stats.index,
    yearly_stats[("Rating", "mean")],
    marker="o",
    label="Average"
)

plt.plot(
    yearly_stats.index,
    yearly_stats[("Rating", "max")],
    marker="s",
    label="Maximum"
)

plt.plot(
    yearly_stats.index,
    yearly_stats[("Rating", "min")],
    marker="^",
    label="Minimum"
)

plt.xlabel("Year")
plt.ylabel("Rating")
plt.title("Rating Statistics by Year")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
